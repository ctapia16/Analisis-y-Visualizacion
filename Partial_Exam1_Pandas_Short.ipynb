## 7) Pandas script (retail_data.csv)
Dataset columns: InvoiceNo, StockCode, Description, Quantity, InvoiceDate, UnitPrice, CustomerID, Country

```
import pandas as pd

# a) Load the dataset
df = pd.read_csv("retail_data.csv")

# b) First 5 rows
df.head()

# c) Last 5 rows
df.tail()

# d) Basic stats for numeric columns
df.describe()

# e) Remove rows with missing values
df_clean = df.dropna()

# f) New column TotalPrice = Quantity * UnitPrice
df_clean = df_clean.copy()
df_clean["TotalPrice"] = df_clean["Quantity"] * df_clean["UnitPrice"]

# Challenge:
# g) Total sales (TotalPrice) by Country
country_total = df_clean.groupby("Country")["TotalPrice"].sum()

# h) Top 10 countries by TotalPrice (descending)
top10 = country_total.sort_values(ascending=False).head(10)
top10
```

## 8) Handling missing values (examples)

```
# a) Detect missing values
df.isnull().sum()

# b) Drop rows with missing values
df_no_missing_rows = df.dropna()

# c) Drop columns with missing values
df_no_missing_cols = df.dropna(axis=1)

df_no_missing_rows.shape, df_no_missing_cols.shape
```

## 9) Selecting rows and columns (examples)

```
# a) Selecting columns
df["UnitPrice"].head()

# b) Selecting rows (first 5 rows by position)
df.iloc[0:5]

# c) Selecting rows and columns (rows 0 to 5, specific columns)
df.loc[0:5, ["Quantity", "UnitPrice", "Country"]]
```

## 10) Graphics

```
import matplotlib.pyplot as plt

# a) Histogram of UnitPrice
plt.figure()
plt.hist(df["UnitPrice"].dropna(), bins=30)
plt.title("UnitPrice Histogram")
plt.xlabel("UnitPrice")
plt.ylabel("Frequency")
plt.show()

# b) Heatmap of correlation matrix
# (Seaborn is convenient for heatmaps)
import seaborn as sns

corr = df.corr(numeric_only=True)

plt.figure()
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
